```python

from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

print(qc.draw())
```

```python

from qiskit_aer import AerSimulator

simulator = AerSimulator()
job = simulator.run(qc, shots=1000)
result = job.result()
counts = result.get_counts()

print("Measurement Results:", counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc_entangled = QuantumCircuit(2, 2)
qc_entangled.h(0)
qc_entangled.cx(0, 1)
qc_entangled.measure([0, 1], [0, 1])

print(qc_entangled.draw())
```

```python

simulator = AerSimulator()
job = simulator.run(qc_entangled, shots=1000)
result = job.result()
counts = result.get_counts()

print("Entanglement Results:", counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc_sandwich = QuantumCircuit(1, 1)
qc_sandwich.h(0)
qc_sandwich.h(0)
qc_sandwich.measure(0, 0)

print("--- CIRCUIT DIAGRAM ---")
print(qc_sandwich.draw())

simulator = AerSimulator()
job = simulator.run(qc_sandwich, shots=1000)
counts = job.result().get_counts()

print("\n--- MEASUREMENT RESULTS ---")
print(counts)
```
