```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

# Metin bazlı çizim (En garantili yöntem!)
print(qc.draw())
```

```python
from qiskit_aer import AerSimulator

# Simülatörü çağırıyoruz
simulator = AerSimulator()
job = simulator.run(qc, shots=1000)
result = job.result()
counts = result.get_counts()

print("Ölçüm Sonuçları:", counts)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1. 2 Kübit ve 2 Klasik Bitlik Devre Oluşturuyoruz
qc_entangled = QuantumCircuit(2, 2)

# 2. 0. kübite Hadamard atıp süperpozisyona sokuyoruz
qc_entangled.h(0)

# 3. CNOT (Controlled-NOT) kapısı ile 0. kübiti 1. kübite dolanık hale getiriyoruz
qc_entangled.cx(0, 1)

# 4. İki kübiti de ölçüyoruz
qc_entangled.measure([0, 1], [0, 1])

# Devreyi ekrana basıyoruz
print(qc_entangled.draw())
```

```python
# Simülatörü çalıştırıyoruz
simulator = AerSimulator()
job = simulator.run(qc_entangled, shots=1000)
result = job.result()
counts = result.get_counts()

print("Dolanıklık Ölçüm Sonuçları:", counts)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1. Tek kübitlik devre
qc_sandwich = QuantumCircuit(1, 1)

# İlk Hadamard: %50-%50 Süperpozisyon oluşturur
qc_sandwich.h(0)

# İkinci Hadamard: Kuantum girişimini başlatır (Dalgaları birleştirir)
qc_sandwich.h(0)

# Ölçüm
qc_sandwich.measure(0, 0)

# Devreyi ekrana basma
print("--- DEVRE ŞEMASI ---")
print(qc_sandwich.draw())

# Simülasyon
simulator = AerSimulator()
job = simulator.run(qc_sandwich, shots=1000)
counts = job.result().get_counts()

print("\n--- ÖLÇÜM SONUÇLARI ---")
print(counts)
```
